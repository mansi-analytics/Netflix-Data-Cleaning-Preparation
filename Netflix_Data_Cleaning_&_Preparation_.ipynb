```python
import pandas as pd
df = pd.read_csv('/content/Dataset.csv')
```

```python
df.head()
```

```python
# column and type
df.info()
```

```python
# checking missing values
print('\n Missing values in each column: ')
print(df.isnull().sum())
```

```python
# checking duplicates
print('\nDuplicate values in each columns: ')
print(df.duplicated().sum())
```

```python
# Check the dataset size
print("\nNumber of rows and columns:", df.shape)
```

```python
# checking data type
print('DATA TYPES')
print(df.dtypes)
```

```python
# checking blank and whitespace only cells
print('BLANK VALUES')
print(
    df.select_dtypes(include='object')
    .apply(lambda col: col.str.strip().eq('').sum())
)
```

```python
# Check common placeholder values
print("\nPLACEHOLDER VALUES")
placeholders = ["Unknown", "unknown", "N/A", "NA", "null", "None"]

for col in df.select_dtypes(include="object").columns:
    count = df[col].isin(placeholders).sum()
    if count > 0:
        print(f"{col}: {count}")
```

```python
# Find rows containing placeholder values in the title column
df[df["title"].isin(
    ["Unknown", "unknown", "N/A", "NA", "null", "None"]
)]
```

```python

# Convert column names to lowercase and replace spaces with underscores
df.columns = (
    df.columns
      .str.strip()
      .str.lower()
      .str.replace(" ", "_")
)

print(df.columns.tolist())
```

```python

# Clean extra spaces in text columns
text_columns = df.select_dtypes(include="object").columns

for col in text_columns:
    df[col] = df[col].str.strip()

print("Text formatting standardized successfully!")
```

```python

# Convert date_added into datetime format
df["date_added"] = pd.to_datetime(
    df["date_added"],
    errors="coerce"
)

print("Data type:", df["date_added"].dtype)
print("Missing dates after conversion:", df["date_added"].isna().sum())
```

```python

# Standardize the type column
df["type"] = df["type"].str.strip().str.title()

# Standardize the rating column
df["rating"] = df["rating"].str.strip().str.upper()

# Standardize country formatting
df["country"] = df["country"].str.strip()

# Check unique values
print("Types:", df["type"].unique())
print("\nRatings:", sorted(df["rating"].dropna().unique()))
print("\nNumber of unique countries:", df["country"].nunique())
```

```python

print("Dataset shape:", df.shape)
print("\nMissing values:")
print(df.isnull().sum())

print("\nDuplicate rows:", df.duplicated().sum())
```

```python

# Inspect duration values
print("Sample duration values:")
print(df["duration"].value_counts().head(15))

# Check duration values that don't match the expected format
duration_text = df["duration"].astype(str).str.strip()

valid_duration = duration_text.str.match(
    r"^\d+\s+(min|season|seasons)$",
    case=False
)

print("\nUnexpected duration values:")
print(df.loc[~valid_duration, "duration"].value_counts())

# Recheck dates
print("\nMissing dates:", df["date_added"].isna().sum())
```

```python

# Check unique show IDs
print("Unique show IDs:", df["show_id"].nunique())
print("Total rows:", len(df))

# Check duplicate IDs
print("Duplicate show IDs:", df["show_id"].duplicated().sum())

# Check data types
print("\nFinal data types:")
print(df.dtypes)

# Check rating categories
print("\nRating counts:")
print(df["rating"].value_counts(dropna=False))

# Final quality check
print("\nMissing values:")
print(df.isnull().sum())
```

```python

# Export the cleaned dataset
output_file = "netflix_cleaned.csv"

df.to_csv(output_file, index=False)

print("Cleaned dataset exported successfully!")
```
